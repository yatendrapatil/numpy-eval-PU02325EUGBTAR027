```python
import numpy as np

np.random.seed(42)
```

```python
A = np.random.randint(1, 11, size=(4, 3))

print("Matrix A:")
print(A)
print("Shape of A:", A.shape)
```

```python
B = np.random.randint(1,11, size=(3,5))

print("Matrix B:")
print(B)
print("Shape of B:", B.shape)
```

```python
assert A.shape[1] == B.shape[0], "Inner dimensions do not match"

print("Dimension validation successful.")
print("A columns:", A.shape[1])
print("B rows:", B.shape[0])
```

```python
C = A @ B

print("Matrix C = A @ B:")
print(C)
print("Shape of C:", C.shape)

assert C.shape == (4, 5)
```

```python
total_sum = np.sum(C)

print("Total sum of elements in C:", total_sum)
```

```python
row_sums = np.sum(C, axis=1)

print("Row-wise sums:")
print(row_sums)
```

```python
column_sums = np.sum(C, axis=0)

print("Column-wise sums:")
print(column_sums)
```
